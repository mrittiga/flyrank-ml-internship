## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Define the evaluation metric for the future ranking experiment.
K = 50

def precision_at_k(actual, predicted, k=50):
    actual = set(actual)
    predicted = predicted[:k]

    if not predicted:
        return 0.0

    return sum(item in actual for item in predicted) / len(predicted)

print("Primary evaluation metric: Precision@50")
print("Evaluation cutoff:", K)
```
